```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

```python
print(iris['target'])
```

```python
pd.DataFrame(iris['data'],columns=iris['feature_names'])
```

```python
#independent features
x=pd.DataFrame(iris['data'],columns=iris['feature_names'])
```

```python
y=iris['target']
```

```python
# train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

in  decission treee you dont require standardization or any kind of normalization

because with help of decission tree we apply the aalgorithm now in order to apply

```python
# Apply Decision Tree
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier()
```

```python
tree.fit(x_train,y_train)
```

```python

# Visualize Decision Tree
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plt.figure(figsize=(15, 10))

plot_tree(tree, filled=True)

plt.show()
```

```python
# Apply Decision Tree
from sklearn.tree import DecisionTreeClassifier

tree_classifier = DecisionTreeClassifier(max_depth=2, random_state=42)
```

```python
tree_classifier.fit(x_train, y_train)
```

```python

# Visualize Decision Tree
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
plt.figure(figsize=(15,10))
plt.show(plot_tree(tree_classifier, filled=True))
```

```python
y_predict = tree_classifier.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
cm=confusion_matrix(y_test,y_predict)
print("confusion matrix:\n",cm)


print("classificattion rreport:\n",classification_report(y_test,y_predict)      )
```

```python
## post prunning good for smaller datastes
```

```python
y_pre=tree.predict(x_test)
```

```python
cm=confusion_matrix(y_test,y_pre)

print("confusion matrix:\n",cm)
print("classificattion rreport:\n",classification_report(y_test,y_pre))
```

# Decision Tree classifier prepruning an dHyperparameter Tunning

```python
param={
    'criterion': ['gini', 'entropy', 'log_loss'],
    'max_depth': [1, 2, 3, 4, 5],
    'splitter': ['best', 'random'],
    'max_features': ['auto', 'sqrt', 'log2']
}
```

```python
param
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
tree_model=DecisionTreeClassifier()
```

```python
grid=GridSearchCV(tree_model,param_grid=param,cv=5,scoring='accuracy')
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_predd=grid.predict(x_test)
```

```python
cm=confusion_matrix(y_test,y_predd)
print("confusion matrix:\n",cm)
print("classificattion rreport:\n",classification_report(y_test,y_predd))
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predd)
```
